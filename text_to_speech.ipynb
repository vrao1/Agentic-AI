```python
import os
from openai import OpenAI
from google.colab import userdata
from pathlib import Path
```

```python
key = userdata.get('OPENAI_API_KEY')
client = OpenAI(api_key=key)
```

```python
current_dir = os.getcwd()
print(current_dir)
```

```python
speech_file_path = os.getcwd() + "/motivational_message.mp3"

with client.audio.speech.with_streaming_response.create(
    model='tts-1',
    voice='nova',
    response_format='mp3',
    input='Hello Future Innovators,As you embark on this exciting journey into the realm of AI development, remember that you are stepping into a field that holds the power to transform the world. Whether you’re crafting algorithms that can make decisions, building models that understand complex patterns, or developing systems that learn and evolve, you are at the frontier of innovation. There will be challenges and puzzles that feel insurmountable at times, but each obstacle is a stepping stone to greater understanding and triumph. Every line of code you write is a building block towards a more intelligent future.Stay curious, because curiosity fuels innovation. Embrace failures as they are simply lessons in disguise. Collaborate and share your knowledge, for it is in the exchange of ideas that we find the brightest solutions.AI is not only about machines learning from data; it’s also about you, the learner, gaining knowledge and insight that empowers you to create, impact, and inspire. Believe in your ability to make a meaningful difference. You have the talent to bring forth innovations that can change lives and improve the world.Keep pushing forward, keep learning, and know that your efforts today are shaping the technologies of tomorrow. The future of AI is bright, and it’s in your capable hands.Dream big, work hard, and let your passion be the spark that lights the path toward your goals.You have got this!'
) as response:
  response.stream_to_file(speech_file_path)
```
